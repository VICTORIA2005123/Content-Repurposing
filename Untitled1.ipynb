````python
def check_facts(content, fact_ledger):
    results = {}

    for key, fact in fact_ledger.items():
        if str(fact).lower() in content.lower():
            results[key] = "PASS"
        else:
            results[key] = "CHECK"

    return results
````

````python
# Combine the thread list into one continuous string for checking
full_thread_content = " ".join(x_thread)
x_check = check_facts(full_thread_content, fact_ledger)

print("X THREAD FACT CHECK")
print("==============================")

for fact, status in x_check.items():
    print(f"{fact}: {status}")
````

````python
import time

# Query Gemini to automatically extract a structured fact ledger with retries and fallback models
prompt = f"""
Analyze this article and extract the main facts into a structured JSON ledger.
We need keys like: organization, event, date, investment, students_supported.

Article:
{article}
"""

# List of models to try in order of preference
models_to_try = ['gemini-3.8-flash', 'gemini-2.5-flash', 'gemini-1.5-flash']
success = False

for model_name in models_to_try:
    if success:
        break
    print(f"Attempting to extract facts using model: {model_name}...")

    # Try up to 3 times with backoff for each model if we encounter rate limits/503s
    for attempt in range(1, 4):
        try:
            response = client.models.generate_content(
                model=model_name,
                contents=prompt
            )
            print("\nAI EXTRACTED FACT LEDGER:")
            print("==============================")
            print(response.text)
            success = True
            break
        except Exception as e:
            err_msg = str(e)
            if "503" in err_msg or "high demand" in err_msg or "429" in err_msg:
                wait_time = attempt * 3
                print(f"  Model {model_name} busy (503/429). Retrying attempt {attempt}/3 in {wait_time}s...")
                time.sleep(wait_time)
            else:
                print(f"  Failed with model {model_name}: {e}")
                break

if not success:
    print("\nFailed to query Gemini after trying multiple models and retrying. Please try again in a few moments.")
````

### 🛠️ Step 5.1: Create the Streamlit Application Script
We will write a file called `app.py` containing the frontend layout and back-end pipeline orchestration logic using the initialized Gemini client.

````python
# 1. Install required packages
!pip install -q gradio google-genai tenacity

import gradio as gr
import json
import re
from google.colab import userdata
from google import genai
from tenacity import retry, stop_after_attempt, wait_exponential

# 2. Initialize Gemini Client using Colab Secrets
GEMINI_API_KEY = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=GEMINI_API_KEY)

# 3. Retry wrapper for API calls
@retry(
    stop=stop_after_attempt(3),
    wait=wait_exponential(multiplier=1, min=2, max=5),
    reraise=True
)
def generate_with_retry(model, contents):
    return client.models.generate_content(model=model, contents=contents)

# 4. Fact Verification Logic
def check_facts_robust(content, fact_ledger):
    results = {}
    content_lower = content.lower()
    for key, fact in fact_ledger.items():
        fact_str = str(fact).strip().lower()
        if fact_str in content_lower:
            results[key] = "✅ PASS"
            continue
        words = re.findall(r'\w+', fact_str)
        important_words = [w for w in words if len(w) > 3 and w not in ['with', 'that', 'this', 'have', 'been']]
        if important_words and all(w in content_lower for w in important_words):
            results[key] = "✅ PASS"
        else:
            results[key] = "❌ CHECK"
    return results

# 5. Core Processing Pipeline
def run_repurposing_chain(article_text):
    if not article_text.strip():
        return "Please enter an article.", "", "", ""

    # Use standard Flash model
    MODEL_NAME = 'gemini-3.8-flash'

    # Stage 1: Extract Ground-Truth Fact Ledger
    ledger_prompt = f"""
    Analyze this article and extract key facts into a valid JSON object with keys:
    organization, event, date, investment, students_supported.
    Do not include markdown backticks.
    Article: {article_text}
    """
    try:
        res1 = generate_with_retry(model=MODEL_NAME, contents=ledger_prompt)
        clean_json = re.sub(r'```json\s*|\s*```', '', res1.text).strip()
        fact_ledger = json.loads(clean_json)
    except Exception as e:
        fact_ledger = {"error": f"Failed to extract facts: {str(e)}"}

    # Stage 2: Content Generation (LinkedIn Post & X Thread)
    gen_prompt = f"""
    Based on this article: "{article_text}" and extracted facts: {json.dumps(fact_ledger)}
    Generate:
    1. A professional LinkedIn post with relevant industry hashtags.
    2. An X (Twitter) thread with 3 tweets numbered 1/3, 2/3, 3/3.

    Format output strictly as JSON:
    {{
      "linkedin": "...",
      "x_thread": ["1/3 ...", "2/3 ...", "3/3 ..."]
    }}
    """
    try:
        res3 = generate_with_retry(model=MODEL_NAME, contents=gen_prompt)
        clean_gen = re.sub(r'```json\s*|\s*```', '', res3.text).strip()
        gen_data = json.loads(clean_gen)
        linkedin_post = gen_data.get("linkedin", "")
        x_thread = gen_data.get("x_thread", [])
    except Exception as e:
        linkedin_post = f"Error generating content: {str(e)}"
        x_thread = []

    # Stage 3: Fact Audit Checks
    linkedin_audit = check_facts_robust(linkedin_post, fact_ledger) if isinstance(fact_ledger, dict) and "error" not in fact_ledger else {}
    x_thread_text = " ".join(x_thread)
    x_audit = check_facts_robust(x_thread_text, fact_ledger) if isinstance(fact_ledger, dict) and "error" not in fact_ledger else {}

    # Stage 4: Formatting Displays
    ledger_display = json.dumps(fact_ledger, indent=2)

    x_display = ""
    for tweet in x_thread:
        x_display += f"- {tweet}\n\n"

    audit_display = "**LinkedIn Audit:**\n\n"
    for k, v in linkedin_audit.items():
        audit_display += f"- `{k}`: {v}\n"

    audit_display += "\n**X Thread Audit:**\n\n"
    for k, v in x_audit.items():
        audit_display += f"- `{k}`: {v}\n"

    return ledger_display, linkedin_post, x_display, audit_display

# 6. Dashboard Layout Setup
default_article = """Marwadi University has launched a new Artificial Intelligence laboratory on September 15, 2026. The university invested ₹50 lakh in the facility. The laboratory is designed to support 200 students and provide practical training in artificial intelligence and machine learning."""

with gr.Blocks(title="AI Content Pipeline") as demo:
    gr.Markdown("# 🔄 AI Content Repurposing Pipeline & Fact Checker")

    with gr.Row():
        # Left Column: Inputs
        with gr.Column(scale=1):
            input_text = gr.Textbox(label="📥 Input Source Article", value=default_article, lines=6)
            submit_btn = gr.Button("⚡ Run Repurposing Pipeline", variant="primary")

        # Right Column: Generated Deliverables & Audits
        with gr.Column(scale=1):
            ledger_output = gr.Code(label="🔍 Extracted Fact Ledger", language="json")
            linkedin_output = gr.Markdown(label="🔗 LinkedIn Deliverable")
            x_output = gr.Markdown(label="🐦 X Thread Deliverable")
            audit_output = gr.Markdown(label="🛡️ Fact Integrity Check Status")

    submit_btn.click(
        fn=run_repurposing_chain,
        inputs=[input_text],
        outputs=[ledger_output, linkedin_output, x_output, audit_output]
    )

demo.launch(share=True, inline=True)
````
